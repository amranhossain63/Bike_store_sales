```python
pearson_corr = bike_data.corr(method='pearson',numeric_only = True)

pearson_corr
```

```python
fig = plt.figure(figsize=(8,8))
plt.matshow(pearson_corr, cmap='RdBu', fignum=fig.number)
plt.xticks(range(len(pearson_corr.columns)), pearson_corr.columns, rotation='vertical');
plt.yticks(range(len(pearson_corr.columns)), pearson_corr.columns);
```

```python
kendall_corr = bike_data.corr(method='kendall',numeric_only = True)

kendall_corr
```

```python
fig = plt.figure(figsize=(8,8))
plt.matshow(kendall_corr, cmap='RdBu', fignum=fig.number)
plt.xticks(range(len(kendall_corr.columns)), kendall_corr.columns, rotation='vertical');
plt.yticks(range(len(kendall_corr.columns)), kendall_corr.columns);
```

```python

spearman_corr = bike_data.corr(method='spearman',numeric_only = True)

spearman_corr
```

```python
fig = plt.figure(figsize=(8,8))
plt.matshow(spearman_corr, cmap='RdBu', fignum=fig.number)
plt.xticks(range(len(spearman_corr.columns)), spearman_corr.columns, rotation='vertical');
plt.yticks(range(len(spearman_corr.columns)), spearman_corr.columns);
```
