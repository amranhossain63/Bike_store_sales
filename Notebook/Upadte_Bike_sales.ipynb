## Relationship between the columns

```python
corr = bike_data.corr(numeric_only = True)
corr
```

```python
fig = plt.figure(figsize=(5,6))
plt.matshow(corr, cmap='RdBu', fignum=fig.number)
plt.xticks(range(len(corr.columns)), corr.columns, rotation='vertical');
plt.yticks(range(len(corr.columns)), corr.columns);
```

```python
pearson_corr = bike_data.corr(method='pearson',numeric_only = True)
pearson_corr
```

```python
fig = plt.figure(figsize = (5,6))
plt.matshow(pearson_corr,cmap = 'seismic',fignum =fig.number )
plt.xticks(range(len(pearson_corr.columns)), pearson_corr.columns, rotation='vertical');
plt.yticks(range(len(pearson_corr.columns)), pearson_corr.columns);
```

```python
kendal_corr = bike_data.corr(method='kendall',numeric_only = True)
kendal_corr
```

```python
fig = plt.figure(figsize =(5,6))
plt.matshow(kendal_corr, cmap = 'RdBu_r',fignum = fig.number)
plt.xticks(range(len(kendal_corr.columns)), kendal_corr.columns,rotation='vertical');
plt.yticks(range(len(kendal_corr.columns)),kendal_corr.columns);
```

```python
spearman_corr = bike_data.corr(method='spearman',numeric_only = True)
spearman_corr
```

```python
fig = plt.figure(figsize = (5,6))
plt.matshow(spearman_corr, cmap='coolwarm',fignum= fig.number)
plt.xticks(range(len(spearman_corr.columns)),spearman_corr.columns, rotation = 'vertical');
plt.yticks(range(len(spearman_corr.columns)), spearman_corr.columns);
```

```python
fig = plt.figure(figsize = (5,6))
plt.matshow(spearman_corr, cmap='PiYG',fignum= fig.number)
plt.xticks(range(len(spearman_corr.columns)),spearman_corr.columns, rotation = 'vertical');
plt.yticks(range(len(spearman_corr.columns)), spearman_corr.columns);
```

## Column wrangling

### Add and calculate a new Revenue_per_Age column

```python
bike_data['Revenue_per_Age'] = bike_data['Revenue'] / bike_data['Customer_Age']
bike_data['Revenue_per_Age'].head()
```

```python
bike_data['Revenue_per_Age'].plot(kind = 'density',figsize = (6,3))
```

```python
bike_data['Revenue_per_Age'].plot(kind = 'hist',figsize = (6,3))
```

### Add and calculate a new Total_Cost column

```python
bike_data['Total_Cost'] = bike_data['Order_Quantity'] * bike_data['Unit_Cost']
bike_data['Total_Cost'].head()
```

```python
(bike_data['Total_Cost'] != bike_data['Cost']).sum()
```

```python
bike_data['Total_Cost'].plot(kind = 'density', figsize = (6,3))
```

```python
bike_data['Total_Cost'].plot(kind = 'hist', figsize = (6,3))
```

#### The relationship between Cost and Profit using a scatter plot

```python
bike_data.plot(kind = 'scatter', x ='Total_Cost', y = 'Profit', figsize = (6,3))
```

### Calculate a new total_Revenue column

```python
bike_data['total_Revenue'] = bike_data['Cost'] + bike_data['Profit']
bike_data['total_Revenue'].head()
```

```python
(bike_data['total_Revenue'] != bike_data['Revenue'] ).sum()
```

```python
bike_data.plot(kind = 'scatter', x ='total_Revenue', y = 'Profit', figsize = (5,3))
```
